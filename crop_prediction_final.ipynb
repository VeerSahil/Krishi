```python
import firebase_admin
from firebase_admin import credentials, firestore

if not firebase_admin._apps:
    cred = credentials.Certificate("smart-agro-ai-fd974-firebase-adminsdk-fbsvc-aa086e000d.json")
    firebase_admin.initialize_app(cred)

db = firestore.client()
```

```python
def predict_crop(entry):
    if entry['rainfall'] > 100:
        return 'Rice'
    else:
        return 'Wheat'
```

```python
def process_latest_entry():
    docs = db.collection('agriculture').order_by('timestamp').limit(1).get()

    for doc in docs:
        data = doc.to_dict()
        crop = predict_crop(data)

        db.collection('agriculture').document(doc.id).update({
            'recommended_crop':crop
        })

        print('Crop predicted:', crop)
```
