# User Input + Rainfall + Soil Nutrients Pipeline

```python
import pandas as pd
import firebase_admin
from firebase_admin import credentials, firestore
from datetime import datetime

cred = credentials.Certificate("smart-agro-ai-fd974-firebase-adminsdk-fbsvc-aa086e000d.json")
firebase_admin.initialize_app(cred)
db = firestore.client()
```

```python
def predict_rainfall(place, season):
    # Placeholder ML logic
    rainfall_map = {
        'Pune': 120,
        'Mumbai': 200,
        'Delhi': 80
    }
    return rainfall_map.get(place, 100)
```

```python
def get_soil_nutrients(soil_type):
    soil_data = pd.DataFrame({
        'soil':['Loamy','Sandy','Clay'],
        'N':[90,40,60],
        'P':[40,20,30],
        'K':[40,20,35]
    })
    row = soil_data[soil_data['soil']==soil_type].iloc[0]
    return row['N'], row['P'], row['K']
```

```python
def push_user_entry(place, season, soil_type):
    rainfall = predict_rainfall(place, season)
    N,P,K = get_soil_nutrients(soil_type)

    data = {
        'place':place,
        'season':season,
        'soil_type':soil_type,
        'rainfall':rainfall,
        'N':int(N),
        'P':int(P),
        'K':int(K),
        'timestamp':datetime.now()
    }

    db.collection('agriculture').add(data)
    print('New Entry Added')
```
