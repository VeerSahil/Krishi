```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
```

```python
rainfall = pd.read_csv("Demo dataset place sang - Copy.csv")
crop = pd.read_csv("Demo dataset crop info.csv")

print("Rainfall Dataset")
display(rainfall.head())

print("Crop Dataset")
display(crop.head())
```

```python
X = rainfall[['Year']]
y = rainfall['Rainfall (mm)']

rain_model = LinearRegression()
rain_model.fit(X,y)
```

```python
predicted_rainfall = rain_model.predict([[2025]])

print("Predicted Rainfall:", predicted_rainfall[0])
```

```python
def rainfall_category(value):
    if value < 650:
        return "Low"
    elif value < 750:
        return "Medium"
    else:
        return "High"

rainfall_level = rainfall_category(predicted_rainfall[0])

print("Rainfall Category:", rainfall_level)
```

```python
le_rain = LabelEncoder()
le_moisture = LabelEncoder()
le_crop = LabelEncoder()

crop['Rainfall'] = le_rain.fit_transform(crop['Rainfall'])
crop['Soil Moisture'] = le_moisture.fit_transform(crop['Soil Moisture'])
crop['Crop'] = le_crop.fit_transform(crop['Crop'])
```

```python
X = crop[['Rainfall','Soil Moisture']]
y = crop['Crop']

crop_model = RandomForestClassifier()

crop_model.fit(X,y)
```

```python
moisture_input = "Medium"

rain_encoded = le_rain.transform([rainfall_level])
moisture_encoded = le_moisture.transform([moisture_input])

prediction = crop_model.predict([[rain_encoded[0], moisture_encoded[0]]])

suggested_crop = le_crop.inverse_transform(prediction)

print("Suggested Crop:", suggested_crop[0])
```

```python

import pickle

# Save Rainfall Model
pickle.dump(rain_model, open("rain_model.pkl", "wb"))

# Save Crop Model
pickle.dump(crop_model, open("crop_model.pkl", "wb"))

# Save Encoders
pickle.dump(le_rain, open("le_rain.pkl", "wb"))
pickle.dump(le_moisture, open("le_moisture.pkl", "wb"))
pickle.dump(le_crop, open("le_crop.pkl", "wb"))

print("All models saved successfully")
```
